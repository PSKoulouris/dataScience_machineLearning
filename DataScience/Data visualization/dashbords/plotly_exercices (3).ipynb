<h1>Airline Data Set</h1>

```python
import requests

def download(url, filename):
    response  = requests.get(url)
    try: 
        if response.status_code == 200:
            print("url successfully reached")
            data = response.content
            with open(filename, "wb") as f:
                f.write(data)
                print(f"Data successfully writen in {filename}")
    except Exception as e:
        print("Error:", e)
```

```python
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv'
filename = "airline_data.csv"
```

```python
download(url, filename)
```

```python
df = pd.read_csv(filename)
df.head()
```

```python
df.shape
```

<h2>Scatter Plot of Ditance Vs Departure Time</h2>

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x = df["Distance"], y = df["DepTime"], mode = "markers", marker = dict(color = "red")))
fig.update_layout(title = "Distance VS Departure Time", xaxis_title = "Distance", yaxis_title = "Departure time")
#fig.show()
```

<h2>Line Plot monthly arrival delay time change over 12 Months </h2>

```python
df_delay = df.groupby(by = "Month")["ArrDelay"].mean().reset_index()
df_delay.head()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x = df_delay["Month"], y = df_delay["ArrDelay"], mode = "lines", marker = dict(color = "green")))
fig.update_layout(title= "Average Departure Delay over 12 Months", xaxis_title = "Months", yaxis_title = "Average departure delay")
#fig.show()
```

<h2>Bar Chart of Number of Flights per Destination</h2>

```python
fig = px.bar(df_destination, x = "DestState", y = "Flights", title = "NUmber of flights to specific destinations")
#fig.show()
```
