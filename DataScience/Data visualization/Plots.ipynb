<h2>Scatter Plot</h2>

<p>trend of total immigrantion to Canada (all countries combined) for the years 1980 - 2013</p>

```python
df_scatter_1 = df.loc[: ,"1980":"2013"].sum(axis = 0)
df_scatter_1 = df_scatter_1.reset_index()
df_scatter_1.columns = ["year", "total"]
df_scatter_1["year"] = df_scatter_1["year"].astype(int)

x = df_scatter_1["year"]
y = df_scatter_1["total"]
fit = np.polyfit(x,y,1)
y_Eq = x*fit[0] + fit[1]
fit
```

```python
df_scatter_1.plot(kind = "scatter", x = "year", y = "total", figsize = (15,8))
plt.plot(x, y_Eq, color = "red")
plt.annotate(f"y={fit[0]:.1f} + {fit[1]:.1f}", xy = (0.6,0.6),xycoords = "axes fraction")
plt.xticks(range(1980, 2014, 5))
plt.title("immigrantion to Canada for the years 1980 - 2013", y = 1.05, fontsize = 25)
```

<p> total immigration from Denmark, Norway, and Sweden to Canada from 1980 to 2013</p>

```python
df_3 = df.loc[["Denmark", "Sweden", "Norway"],"1980":"2013"]
df_3_1 = df_3.sum(axis = 0).to_frame()
df_3_1 = df_3_1.reset_index()
df_3_1.columns = ["year","total"]
df_3_1["year"] = df_3_1["year"].astype(int)

x = df_3_1["year"]
y = df_3_1["total"]
fit = np.polyfit(x,y, deg = 1)

def eq(x):
    return fit[0]*x + fit [1]
```

```python
df_3_1.plot(kind = "scatter", x = "year", y = "total", figsize = (15,7))
plt.xticks(range(1980,2013,5))
plt.ylim(0,)
plt.title("cummulative immigrantion from Sweden, Denmark, and Norway to Canada for the years 1980 - 2013", fontsize = 15, y = 1.05)

plt.plot(x, eq(x), color = "red")
plt.annotate(f"y = {fit[0]:.1f}x + {fit[1]:.1f}",xy = (0.05,0.67), xycoords="axes fraction")
```

<h2>Bubble Plot</h2>

<p>effect of this crisis, and compare Argentina's immigration to that of it's neighbour Brazil. Let's do that using a bubble plot of immigration from Brazil and Argentina for the years 1980 - 2013. We will set the weights for the bubble as the normalized value of the population for each year</p>

```python
df_1 = df.loc[:, "1980":"2013"]
df_1 = df_1.transpose()
df_1.index.name = "Year"
df_1.reset_index(inplace = True)
df_1["Year"] = df_1["Year"].astype(int)
df_1.head()
```

<h3>Rescaling (min-max normalization)</h3>
<p>Also known as min-max scaling or min-max normalization, consists in rescaling the range of features to scale the range in [0, 1] or [−1, 1]. Selecting the target range depends on the nature of the data. The general formula for a min-max of [0, 1] is given as:

<p>
\( x' = \frac{x - \min(x)}{\max(x) - \min(x)} \)
</p>
</p>

```python
normalized_brazil = ((df_1["Brazil"] - df_1["Brazil"].min() )/ (df_1["Brazil"].max() - df_1["Brazil"].min()))
normalized_argentina = ((df_1["Argentina"] - df_1["Argentina"].min()) / (df_1["Argentina"].max() - df_1["Argentina"].min()))
normalized_brazil
```

```python
ax0 = df_1.plot(kind = "scatter", x = "Year", y = "Brazil", figsize = (15,8),  s=normalized_brazil * 2000 + 10, alpha = .5)  # pass in weights 
ax1 = df_1.plot(kind = "scatter", x = "Year", y = "Argentina", color = "red", ax = ax0,  s=normalized_argentina * 2000 + 10, alpha = .5)

ax0.set_title("Immigration from Brazil and Argentina to Canada between 1980 and 2013", fontsize = 20, y = 1.05)
ax0.legend(["Brazil", "Argentina"], loc = "upper left", fontsize = 15)
ax0.set_ylabel("number of immigrants")
```

<p>Immigration from China and India to Canada from 1980 to 2013</p>

```python
normalized_china = ((df_1["China"] - df_1["China"].min() )/ (df_1["China"].max() - df_1["China"].min()))
normalized_india = ((df_1["India"] - df_1["India"].min()) / (df_1["India"].max() - df_1["India"].min()))
normalized_china
```

```python
ax0 = df_1.plot(kind = "scatter", x = "Year", y = "China", figsize = (15,8),  s=normalized_china * 2000 + 10, alpha = .5)  # pass in weights 
ax1 = df_1.plot(kind = "scatter", x = "Year", y = "India", color = "red", ax = ax0,  s=normalized_india * 2000 + 10, alpha = .5)

ax0.set_title("Immigration from China and India to Canada between 1980 and 2013", fontsize = 20, y = 1.05)
ax0.legend(["China", "India"], loc = "upper left", fontsize = 15)
ax0.set_ylabel("number of immigrants")
```
