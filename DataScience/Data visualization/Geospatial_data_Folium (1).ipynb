```python
import requests

def download(url, filename):
    response = requests.get(url)
    try:
        if response.status_code == 200:
            print("url accesssed successfully")
            data = response.content
            with open(filename, "wb") as f:
                f.write(data)
                print("data successfully downloaded")
    except Exception as e:
        print("Error:", e)
```

<h2>San Francisco Incident Map </h2>

```python
latitude = 37.77
longitude = -122.42
```

```python
sf_map = folium.Map(location = [latitude, longitude], zoom_start = 12)
#sf_map 
```

```python
df_incidents = df.head(300)

incidents = folium.map.FeatureGroup()

for lat,long in zip(df_incidents["Y"], df_incidents["X"]):
    incidents.add_child(
    folium.vector_layers.CircleMarker(
    [lat,long],
    radius = 5,
    color = "blue",
    fill = True,
    fill_color = "yellow",
    fill_opacity = .6
        )
    )
  
longitude_data = list(df_incidents["X"])
latitude_data = list(df_incidents["Y"])
labels = list(df_incidents["Category"])

for lon, lat, label in zip(longitude_data, latitude_data, labels):
    folium.Marker([lat,lon], popup = label).add_to(sf_map)

    
    
sf_map.add_child(incidents)
#sf_map
```

```python
df_incidents = df.head(300)

latitude = 37.77
longitude = -122.42

sf_map = folium.Map(location = [latitude, longitude], zoom_start = 12)

incidents = folium.map.FeatureGroup()

for lat,lon,label in zip(df_incidents["Y"], df_incidents["X"], df_incidents["Category"]):
    incidents.add_child(
    folium.vector_layers.CircleMarker(
    [lat,lon],
    radius = 5,
    color = "blue",
    fill = True,
    fill_color = "yellow",
    fill_opacity = .6,
    popup = label
        )
    )
    
sf_map.add_child(incidents)
#sf_map
```

```python
from folium import plugins

latitude = 37.77
longitude = -122.42

sf_map = folium.Map(location=[latitude, longitude], zoom_start = 12)

incidents_cluster = plugins.MarkerCluster().add_to(sf_map)

for lan, lon, label in zip(df["Y"], df["X"], df["Category"]):
    folium.Marker(
    location = [lan, lon],
    icon = None,
    popup = label
    ).add_to(incidents_cluster)

#sf_map
```
