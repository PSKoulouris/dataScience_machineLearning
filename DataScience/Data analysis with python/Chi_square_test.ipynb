```python
import pandas as pd
from scipy.stats import chi2_contingency
import numpy as np
```

```python
data_male = {"like":20,
             "dislike": 30}
data_femelle = {"like" : 25,
                "dislike" : 25}
```

```python
data = np.array([
                [data_male["like"],data_male["dislike"]],
                [data_femelle["like"], data_femelle["dislike"]]
                ])
```

```python
data
```

```python
df = pd.DataFrame(data, columns = ["Like", "Dislike"], index = ["male", "femelle"])
df
```

```python
from IPython.display import display, HTML
chi, p, dof, expected = chi2_contingency(df)
df = pd.DataFrame(expected, index = ["male", "femelle"], columns = ["like", "dislike"])
display(HTML(f"""
        <h3>Chi-Square Test of Independence</h3>
        <ul style="list-style-type:none;">
        <li> Chi_square : {chi} </li>
        <li> Degree of freedom : {dof} </li>
        <li> p-value : {p} </li>
        <br>
        {df.to_html(index = True)}
        <br>
        <b>result</b> : χ²({dof}) = {chi:.2f}, <i>p</i> = {p:.2f}"""
))
```
