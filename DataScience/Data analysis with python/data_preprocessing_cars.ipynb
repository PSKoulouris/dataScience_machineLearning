```python
from pyodide.http import pyfetch
```

```python
file_path="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/auto.csv"
```

```python
async def download(file_path, filename):
    response = await pyfetch(file_path)
    if response.status == 200:
        print("connection successful")
        with open (filename, "wb") as f:
            f.write(await response.bytes())
        print(f"{filename} downloaded successfully")
```

```python
await download(file_path, "usedcars.csv")
```

```python
csvFile = "usedcars.csv"
headers= ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]

df = pd.read_csv(csvFile, names = headers)

df.head()
```

```python
import numpy as np

df1 = df.replace("?", np.nan)
df1.describe(include = "all")
```

```python
(df1 == "?").sum()
```

```python
df1.isnull().sum()
```

```python
avg_normalized_losses = df1["normalized-losses"].astype("float").mean(axis = 0)
avg_bore = df1["bore"].astype("float").mean(axis = 0)
avg_stroke = df1["stroke"].astype("float").mean(axis = 0)
avg_horsepower = df1["horsepower"].astype("float").mean(axis = 0)
avg_peak_rpm = df1["peak-rpm"].astype("float").mean(axis = 0)

print(avg_normalized_losses, 
avg_bore,
avg_stroke,
avg_horsepower,
avg_peak_rpm)
```

```python
df1["normalized-losses"] = df1["normalized-losses"].replace(np.nan, avg_normalized_losses)
df1["bore"] = df1["bore"].replace(np.nan, avg_bore)
df1["stroke"] = df1["stroke"].replace(np.nan, avg_stroke)
df1["horsepower"] = df1["horsepower"].replace(np.nan, avg_horsepower)
df1["peak-rpm"] = df1["peak-rpm"].replace(np.nan, avg_peak_rpm)
```

```python
df1.isnull().sum()
```

```python
#frequency_doors_max = df1["num-of-doors"].value_counts().idxmax()
mode_doors = df1["num-of-doors"].mode()[0]
mode_doors
```

```python
df1["num-of-doors"] = df1["num-of-doors"].fillna(mode_doors)
```

```python
df1 = df1.dropna(subset = ["price"])
```

```python
df1.reset_index(drop = True, inplace = True)
```

```python
start_value = min(df2["horsepower"])
end_value = max(df2["horsepower"])
delimiters = 4
bins = np.linspace(start_value, end_value, delimiters)
```

```python
df2["binned_horsepower"] = pd.cut(df2["horsepower"], bins, labels = bin_names)
```

```python
dummies_variable_fuel = pd.get_dummies(df2["fuel-type"])
```

```python
dummies_variable_fuel.rename(columns={"diesel":"fuel-type-diesel","gas":"fuel-type-gas"}, inplace = True)
```

```python
df_final = pd.concat([df2, dummies_variable_fuel], axis = 1)
df_final.drop("fuel-type", axis = 1, inplace = True)
df_final.head()
```

```python
dummie_variables_aspiration = pd.get_dummies(df_final["aspiration"])
dummie_variables_aspiration.rename(columns={"std":"aspiration-standard","turbo":"apiration-turbo"}, inplace = True)
dummie_variables_aspiration.head()
```

```python
df_final = pd.concat([df_final, dummie_variables_aspiration], axis = 1)
df_final.head()
```

```python
df_final = df_final.drop(df_final.columns[-4:-3], axis = 1)
df_final = df_final.drop("aspiration", axis = 1)
df_final.head()
```
